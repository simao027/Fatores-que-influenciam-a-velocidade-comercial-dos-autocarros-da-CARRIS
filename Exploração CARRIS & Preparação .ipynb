```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df_PGIL_CARRIS_2023 = df_PGIL_CARRIS_2023.drop(columns=['last_updated'])

df_PGIL_CARRIS_2023 = df_PGIL_CARRIS_2023.drop(columns=['route_id'])

df_PGIL_CARRIS_2023 = df_PGIL_CARRIS_2023.drop(columns=['entity_location'])

df_PGIL_CARRIS_2023 = df_PGIL_CARRIS_2023.drop(columns=['position'])

df_PGIL_CARRIS_2023['entity_ts'] = pd.to_datetime(df_PGIL_CARRIS_2023['entity_ts'], errors='coerce')

df_PGIL_CARRIS_2023['status'] = df_PGIL_CARRIS_2023['status'].astype('category')

df_PGIL_CARRIS_2023['entity_id'] = df_PGIL_CARRIS_2023['entity_id'].str.extract(r'\.(\d+)$')[0].astype('int64')
```

```python
data_inicio = pd.to_datetime('2023-09-01')
data_fim = pd.to_datetime('2023-10-31')


fora_intervalo = df_PGIL_CARRIS_2023[
    (df_PGIL_CARRIS_2023['entity_ts'].dt.date < data_inicio.date()) | 
    (df_PGIL_CARRIS_2023['entity_ts'].dt.date > data_fim.date())
]

print(f"Total de linhas com data fora do intervalo de 1 de setembro a 31 de outubro de 2023: {len(fora_intervalo)}")


datas_fora = fora_intervalo['entity_ts'].dt.date.drop_duplicates().sort_values()
print("\nDatas encontradas fora do intervalo:")
print(datas_fora.tolist())
```

```python
df_PGIL_CARRIS_2023 = df_PGIL_CARRIS_2023[~df_PGIL_CARRIS_2023['entity_ts'].dt.date.isin(datas_fora)]

fora_intervalo = df_PGIL_CARRIS_2023[
    (df_PGIL_CARRIS_2023['entity_ts'].dt.date < data_inicio.date()) | 
    (df_PGIL_CARRIS_2023['entity_ts'].dt.date > data_fim.date())
]

print(f"Total de linhas com data fora do intervalo de 1 de setembro a 31 de outubro de 2023: {len(fora_intervalo)}")
```

```python
def normalizar_route_number(x):
    if pd.isna(x):
        return None
    try:
        
        return str(int(float(x)))
    except:
        
        return str(x).strip()

df_PGIL_CARRIS_2023['route_number'] = df_PGIL_CARRIS_2023['route_number'].apply(normalizar_route_number)
```

## 3.1.2.5 Paragens

```python
n_paragens_diferentes = df_PGIL_CARRIS_2023['stop_id'].nunique()
print(f"Número de paragens diferentes: {n_paragens_diferentes}")
```

```python

locais_por_paragem = df_PGIL_CARRIS_2023.groupby('stop_id').apply(
    lambda x: x[['latitude', 'longitude']].drop_duplicates().shape[0]
).reset_index(name='n_locais_diferentes')


paragens_duvidosas = locais_por_paragem[locais_por_paragem['n_locais_diferentes'] > 1]

print(f"Número de stop_id com mais de 1 localização: {len(paragens_duvidosas)}")
print(paragens_duvidosas.head(10))
```

```python
# Passar todos os stop_id para strings e remover zeros à esquerda, pontos e espaços
stop_ids_limpos = (
    df_PGIL_CARRIS_2023['stop_id']
    .astype(str)
    .str.strip()
    .str.replace(r'\.0$', '', regex=True)  # remover .0 do fim
    .str.lstrip('0')  # remover zeros à esquerda
)


stop_ids_limpos_unique = stop_ids_limpos.nunique()
print(f"Stop IDs únicos após limpeza: {stop_ids_limpos_unique}")


stop_ids_originais_unique = df_PGIL_CARRIS_2023['stop_id'].nunique()
print(f"Stop IDs únicos originais: {stop_ids_originais_unique}")

```

```python

locais_2 = df_PGIL_CARRIS_2023[df_PGIL_CARRIS_2023['stop_id'] == 2.0][['latitude', 'longitude']]


print(f"Número de localizações únicas da paragem 2.0: {locais_2.drop_duplicates().shape[0]}")

print(locais_2.value_counts().head(10))
```

```python
# 1. Número médio de paragens diferentes por rota por dia
paragens_por_rota_dia = (
    df_PGIL_CARRIS_2023
    .assign(data=lambda df: df['entity_ts'].dt.date)
    .groupby(['data', 'route_number'])['stop_id']
    .nunique()
    .reset_index(name='n_paragens_diferentes')
)
media_paragens_por_rota_dia = paragens_por_rota_dia['n_paragens_diferentes'].mean()
print(f"Número médio de paragens diferentes por rota por dia: {media_paragens_por_rota_dia:.2f}")

# 2. Diferenças por dia da semana (1=Segunda, 7=Domingo)
paragens_por_dia_semana = (
    df_PGIL_CARRIS_2023
    .assign(
        data=lambda df: df['entity_ts'].dt.date,
        dia_da_semana=lambda df: df['entity_ts'].dt.dayofweek + 1
    )
    .groupby(['dia_da_semana', 'data', 'route_number'])['stop_id']
    .nunique()
    .reset_index()
)
media_por_semana = paragens_por_dia_semana.groupby('dia_da_semana')['stop_id'].mean()
print("\nMédia de paragens diferentes por rota por dia da semana (1=Segunda, 7=Domingo):")
print(media_por_semana)

# 3. Diferenças por hora do dia
paragens_por_hora = (
    df_PGIL_CARRIS_2023
    .assign(
        hora=lambda df: df['entity_ts'].dt.hour,
        data=lambda df: df['entity_ts'].dt.date
    )
    .groupby(['hora', 'data', 'route_number'])['stop_id']
    .nunique()
    .reset_index()
)
media_por_hora = paragens_por_hora.groupby('hora')['stop_id'].mean()
print("\nMédia de paragens diferentes por rota por hora do dia:")
print(media_por_hora)
```

```python
# Obter, para cada trip, a sequência de stop_ids e respetiva localização do primeiro registo
df_sorted = df_PGIL_CARRIS_2023.sort_values(['trip_id', 'entity_ts'])
primeiros_stops = (
    df_sorted
    .groupby(['trip_id', 'stop_id'], as_index=False)
    .first()
)

# Agora, para cada trip, detetar apenas as transições de stop_id (primeira vez que aparece em cada trip)
primeiros_stops_trip = (
    primeiros_stops
    .sort_values(['trip_id', 'entity_ts'])
    .groupby('trip_id')
    .apply(lambda x: x.loc[x['stop_id'].shift() != x['stop_id']])
    .reset_index(drop=True)
)

# Contar as paragens mais usadas (considerando localização do primeiro registo)
top_paragens = (
    primeiros_stops_trip
    .groupby(['stop_id', 'latitude', 'longitude'])
    .size()
    .reset_index(name='n_passagens')
    .sort_values('n_passagens', ascending=False)
    .head(10)
)
print("Top 10 paragens mais usadas (primeiro registo de cada transição):")
display(top_paragens)
```

```python

stopids_por_rota = df_PGIL_CARRIS_2023.groupby('route_number')['stop_id'].nunique().sort_values(ascending=False)
print("Rotas com mais stop_ids diferentes (global):")
print(stopids_por_rota.head(10))


stopids_por_rota_dia = (
    df_PGIL_CARRIS_2023
    .assign(data=lambda df: df['entity_ts'].dt.date)
    .groupby(['data', 'route_number'])['stop_id']
    .nunique()
    .reset_index(name='n_stopids_dia')
)
media_stopids_por_rota = (
    stopids_por_rota_dia.groupby('route_number')['n_stopids_dia']
    .mean()
    .sort_values(ascending=False)
)
print("\nTop 10 rotas com maior número médio diário de stop_ids diferentes:")
print(media_stopids_por_rota.head(10))

evolucao_horaria_global = (
    df_PGIL_CARRIS_2023
    .assign(hora=lambda df: df['entity_ts'].dt.hour)
    .groupby('hora')['stop_id']
    .nunique()
)
plt.figure(figsize=(12,6))
plt.plot(evolucao_horaria_global.index, evolucao_horaria_global.values, marker='o')
plt.title("Evolução horária do número de stop_ids diferentes (todas as rotas)")
plt.xlabel("Hora do dia")
plt.ylabel("N.º de stop_ids diferentes")
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
# 1. Top 10 dias com maior número de stop_ids diferentes
stopids_por_dia = (
    df_PGIL_CARRIS_2023
    .assign(data=lambda df: df['entity_ts'].dt.date)
    .groupby('data')['stop_id']
    .nunique()
    .sort_values(ascending=False)
)
print("Top 10 dias com maior número de stop_ids diferentes:")
for data, n in stopids_por_dia.head(10).items():
    print(f"{data}: {n} stop_ids diferentes")

# 2. Gráfico da evolução semanal do número de stop_ids diferentes por dia
plt.figure(figsize=(14,5))
stopids_por_dia.plot(marker='o')
plt.title("Evolução diária do número de stop_ids diferentes (Set-Out 2023)")
plt.xlabel("Data")
plt.ylabel("N.º de stop_ids diferentes")
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

## 3.1.2.6	Extensão Geográfica das Rotas

```python
from datetime import timedelta

# Acrescentar data da viagem
df = df_PGIL_CARRIS_2023.copy()
df['data_viagem'] = df['entity_ts'].dt.date

# Ordenar por sequência temporal
df = df.sort_values(['entity_id', 'trip_id', 'data_viagem', 'entity_ts'])

# Calcular tempo entre registos consecutivos dentro do mesmo trip_id
df['delta_t'] = df.groupby(['entity_id', 'trip_id', 'data_viagem'])['entity_ts'].diff().dt.total_seconds().fillna(0)

# Marcar nova subviagem se a diferença temporal for > 15 minutos (900 segundos)
df['nova_subviagem'] = (df['delta_t'] > 900).astype(int)

# Gerar sub_viagem_id cumulativo dentro de cada trip_id
df['sub_viagem_id'] = df.groupby(['entity_id', 'trip_id', 'data_viagem'])['nova_subviagem'].cumsum()

# Agrupar por subviagem identificada
chave_viagem = ['entity_id', 'trip_id', 'data_viagem', 'sub_viagem_id']
viagens_info = []

def haversine(lat1, lon1, lat2, lon2):
    R = 6371
    lat1, lon1, lat2, lon2 = map(np.radians, [lat1, lon1, lat2, lon2])
    dlat = lat2 - lat1
    dlon = lon2 - lon1
    a = np.sin(dlat/2)**2 + np.cos(lat1)*np.cos(lat2)*np.sin(dlon/2)**2
    c = 2 * np.arcsin(np.sqrt(a))
    return R * c

for chave, grupo in df.groupby(chave_viagem):
    grupo = grupo.sort_values('entity_ts')
    rota = grupo['route_number'].iloc[0]
    inicio = grupo['entity_ts'].min()
    fim = grupo['entity_ts'].max()
    extensao = 0

    if len(grupo) > 1:
        lat1 = grupo['latitude'].values[:-1]
        lon1 = grupo['longitude'].values[:-1]
        lat2 = grupo['latitude'].values[1:]
        lon2 = grupo['longitude'].values[1:]
        extensao = haversine(lat1, lon1, lat2, lon2).sum()

    viagens_info.append({
        'entity_id': chave[0],
        'trip_id': chave[1],
        'data_viagem': chave[2],
        'sub_viagem_id': chave[3],
        'rota': rota,
        'extensao_km': extensao,
        'inicio': inicio,
        'fim': fim
    })

# Construir DataFrame com métricas por subviagem
df_trips = pd.DataFrame(viagens_info)
df_trips['tempo_horas'] = (df_trips['fim'] - df_trips['inicio']).dt.total_seconds() / 3600

# Aplicar filtros de qualidade
df_trips_filtrado = df_trips[
    (df_trips['extensao_km'] > 0.5) &
    (df_trips['extensao_km'] <= 50) &
    (df_trips['tempo_horas'] > 0.2)
]

# Estatísticas por rota, já com viagens reais e limpas
df_rotas_ext_melhorado = (
    df_trips_filtrado.groupby('rota')
    .agg(
        extensao_mediana_km=('extensao_km', 'median'),
        extensao_p90_km=('extensao_km', lambda x: x.quantile(0.90)),
        extensao_media_km=('extensao_km', 'mean'),
        extensao_total_km=('extensao_km', 'sum'),
        n_trips=('sub_viagem_id', 'count')  
    )
    .reset_index()
    .sort_values('extensao_mediana_km', ascending=False)
)
```

```python
from typing import List
import logging
from scipy.spatial import cKDTree

# CONFIGURAÇÃO GERAL
CONFIG = {
    "VMC_MAX_KMH": 80.0,
    "VELOCIDADE_MIN_KMH": 2.0,
    "EXT_MIN_KM": 0.5,
    "EXT_MAX_KM": 50.0,
    "TEMP_MIN_H": 0.2,
    "HORAS_PICO": [(6, 10), (16, 20)],  # 6h-9h59 e 16h-19h59
    "SUBVIAGEM_BREAK_S": 900,  # 15 minutos em segundos
    "COLUNAS_FEATURES": [
        'dia_semana', 'classe_hora', 'feriado', 'zona_id',
        'classe_chuva', 'atraso_medio_waze', 'route_number'
    ],
    "COLUNA_TARGET": 'vmc_kmh'
}
logging.basicConfig(level=logging.INFO, format='%(levelname)s: %(message)s')

# FUNÇÃO Haversine
def haversine(lat1: np.ndarray, lon1: np.ndarray,
              lat2: np.ndarray, lon2: np.ndarray) -> np.ndarray:
    R = 6371.0
    φ1, λ1, φ2, λ2 = map(np.radians, [lat1, lon1, lat2, lon2])
    dφ = φ2 - φ1
    dλ = λ2 - λ1
    a = np.sin(dφ / 2) ** 2 + np.cos(φ1) * np.cos(φ2) * np.sin(dλ / 2) ** 2
    c = 2 * np.arcsin(np.sqrt(a))
    return R * c

# IMUTAR ATRASO MÉDIO WAZE
def imputar_atraso_waze(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    grupo1 = df.groupby(['zona_id', 'dia_semana', 'hora', 'feriado'])['atraso_medio_waze'].transform('mean')
    df['atraso_medio_waze'] = df['atraso_medio_waze'].fillna(grupo1)
    grupo2 = df.groupby(['zona_id', 'hora', 'feriado'])['atraso_medio_waze'].transform('mean')
    df['atraso_medio_waze'] = df['atraso_medio_waze'].fillna(grupo2)
    grupo3 = df.groupby(['zona_id', 'hora'])['atraso_medio_waze'].transform('mean')
    df['atraso_medio_waze'] = df['atraso_medio_waze'].fillna(grupo3)
    grupo4 = df.groupby(['zona_id'])['atraso_medio_waze'].transform('mean')
    df['atraso_medio_waze'] = df['atraso_medio_waze'].fillna(grupo4)
    df['atraso_medio_waze'] = df['atraso_medio_waze'].fillna(df['atraso_medio_waze'].mean())
    return df

# CRIAR VARIÁVEIS TEMPORAIS E DETECTAR SUBVIAGENS
def preparar_dados_temporais_e_subviagens(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    df['hora'] = df['entity_ts'].dt.hour
    df['dia_semana'] = df['entity_ts'].dt.weekday
    df['feriado'] = False
    df['data_viagem'] = df['entity_ts'].dt.date
    df = df.sort_values(['entity_id', 'trip_id', 'data_viagem', 'entity_ts'])
    df['delta_t'] = df.groupby(['entity_id', 'trip_id', 'data_viagem'])['entity_ts'].diff().dt.total_seconds().fillna(0)
    df['nova_subviagem'] = (df['delta_t'] > CONFIG["SUBVIAGEM_BREAK_S"]).astype(int)
    df['sub_viagem_id'] = df.groupby(['entity_id', 'trip_id', 'data_viagem'])['nova_subviagem'].cumsum()
    return df

# FILTRAGEM DE VIAGENS COMPLETAS ROBUSTAS
def filtrar_viagens_robustas(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    chave = ['entity_id', 'trip_id', 'data_viagem', 'sub_viagem_id']
    viagens_meta = []
    for key, grupo in df.groupby(chave):
        grupo = grupo.sort_values('entity_ts')
        if len(grupo) < 2:
            continue
        inicio = grupo['entity_ts'].iloc[0]
        fim = grupo['entity_ts'].iloc[-1]
        tempo_h = (fim - inicio).total_seconds() / 3600
        lat1 = grupo['latitude'].values[:-1]
        lon1 = grupo['longitude'].values[:-1]
        lat2 = grupo['latitude'].values[1:]
        lon2 = grupo['longitude'].values[1:]
        extensao = haversine(lat1, lon1, lat2, lon2).sum()
        velocidade = extensao / tempo_h if tempo_h > 0 else 0
        viagens_meta.append((key, extensao, tempo_h, velocidade))
    valid_keys = {
        key for key, ext, tempo, vel in viagens_meta
        if (CONFIG["EXT_MIN_KM"] < ext <= CONFIG["EXT_MAX_KM"]
            and tempo > CONFIG["TEMP_MIN_H"]
            and CONFIG["VELOCIDADE_MIN_KMH"] < vel <= CONFIG["VMC_MAX_KMH"])
    }
    antes = df.shape[0]
    df_filtrado = df[df.set_index(chave).index.isin(valid_keys)].copy()
    logging.info(f"Filtragem viagens: de {antes} registos passaram para {df_filtrado.shape[0]}")
    return df_filtrado

# CÁLCULO DA VMC POR SEGMENTO
def calcular_vmc_robusto(df: pd.DataFrame, vmc_max: float) -> pd.DataFrame:
    df = df.copy()
    df = df.sort_values(['entity_id', 'trip_id', 'data_viagem', 'sub_viagem_id', 'entity_ts'])
    grupo = df.groupby(['entity_id', 'trip_id', 'data_viagem', 'sub_viagem_id'])
    df['lat_next'] = grupo['latitude'].shift(-1)
    df['lon_next'] = grupo['longitude'].shift(-1)
    df['ts_next'] = grupo['entity_ts'].shift(-1)
    df = df.dropna(subset=['lat_next', 'ts_next']).copy()
    df['dist_km'] = haversine(df['latitude'], df['longitude'], df['lat_next'], df['lon_next'])
    df['time_h'] = (df['ts_next'] - df['entity_ts']).dt.total_seconds() / 3600
    df = df.query("dist_km > 0 and time_h > 0")
    df['vmc_kmh'] = df['dist_km'] / df['time_h']
    antes = df.shape[0]
    df = df.query("vmc_kmh <= @vmc_max")
    logging.info(f"Filtragem segmentos (> VMC_MAX): de {antes} passaram para {df.shape[0]}")
    return df

# Função para classificar hora em 'Pico' ou 'Não Pico' usando a config
def classificar_pico(hora):
    for inicio, fim in CONFIG["HORAS_PICO"]:
        if inicio <= hora < fim:
            return 'Pico'
    return 'Não Pico'

# CRIAR FEATURES PARA MODELAR
def criar_features_modelo(df: pd.DataFrame) -> pd.DataFrame:
    df = df.copy()
    if 'hora' not in df.columns:
        df['hora'] = df['entity_ts'].dt.hour
    df['classe_hora'] = df['hora'].apply(classificar_pico)
    return df

# FINALIZAR DATASET PARA MODELAR
def finalizar_dataset_modelo(df: pd.DataFrame, features: List[str], target: str) -> pd.DataFrame:
    cols = features + [target]
    df_final = df[cols].dropna().copy()
    logging.info(f"Dataset final para modelação: {df_final.shape[0]} registos, {len(cols)} colunas")
    return df_final

# PIPELINE COMPLETO
df_modelo = (
    df_completo
    .pipe(preparar_dados_temporais_e_subviagens)
    .pipe(filtrar_viagens_robustas)
    .pipe(imputar_atraso_waze)  # Imputação espacial e temporal
    .pipe(calcular_vmc_robusto, vmc_max=CONFIG["VMC_MAX_KMH"])
    .pipe(criar_features_modelo)
    .pipe(finalizar_dataset_modelo, features=CONFIG["COLUNAS_FEATURES"], target=CONFIG["COLUNA_TARGET"])
)

display(df_modelo.head())
df_modelo.info()
```
